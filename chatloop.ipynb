```python
import os 
from openai import AzureOpenAI
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
client=AzureOpenAI(
    api_key=os.environ["AZURE_GPT_5_MINI_API_KEY"],
    azure_endpoint=os.environ["AZURE_ENDPOINT"],
    api_version="2024-08-01-preview"

)
```

```python
messages=[{"role":"system","content":"You are a helpful assistant"}]
```

```python
while True:
    user=input("Hey, I am Shubham's personal AI assistant. Ask me anything: ")
    if user.strip().casefold()=="quit":
        break
    if not user.strip():
        continue
    messages.append({"role":"user","content":user})
    response=client.chat.completions.create(messages=messages,model="gpt-5-mini")
    reply=response.choices[0].message.content
    messages.append({"role":"assistant","content":reply})
    print(reply)
```
