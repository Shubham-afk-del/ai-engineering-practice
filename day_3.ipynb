```python
import os 
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
model="qwen2.5:3b-instruct-q4_0"
```

```python
client=OpenAI(api_key="ollama",base_url="http://localhost:11434/v1")
```

```python
message = [
    {"role": "system", "content": "You are a waiter at a chai stall. Answer in one short line."},
    {"role": "user", "content": "Ek chai kitne ki hai?"},
    {"role": "assistant", "content": "Ek chai 10 rupees ki hai."},
    {"role": "user", "content": "Toh 3 chai kitne ki?"},
]
r = client.chat.completions.create(model=model, messages=message, temperature=0)
print(r.choices[0].message.content)
```

```python
message=[{"role":"user","content":"hi my name is shubham whats your name and what you can do "}]
stream=client.chat.completions.create(model=model,messages=message,stream=True)
for chunk in stream:
    print(chunk.choices[0].delta.content or "",end="")
```
